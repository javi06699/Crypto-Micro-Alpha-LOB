```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import polars as pl
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from hyperopt import fmin, tpe, hp, STATUS_OK, Trials

# 1. VERIFICACIÓN DE DISPOSITIVO
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🖥️ Hyperopt ejecutándose sobre: {device}")

# =====================================================================
# 2. PREPARACIÓN DE MATRICES Y SPLIT PARA VALIDACIÓN CRUZADA TEMPORAL
# =====================================================================
target_col = "label_k_100"
all_columns = dataset_final.collect_schema().names()
feature_cols = [
    c for c in all_columns if c.startswith("v") ] + ["OBI_Nivel_1"]

# Cargamos el dataset completo ordenado cronológicamente
df_all = dataset_final.collect()
if "timestamp" in df_all.columns:
    df_all = df_all.sort("timestamp")

# Aislamos un 15% al final como TEST CIEGO definitivo (fuera de Hyperopt)
total_rows = len(df_all)
test_cutoff = int(total_rows * 0.85)

df_cv_pool = df_all.slice(0, test_cutoff)  # Pool para la validación cruzada
df_final_test = df_all.slice(test_cutoff)  # Test ciego final

# =====================================================================
# 3. CREACIÓN DE LOS FOLDS TEMPORALES (Directamente en CUDA para velocidad)
# =====================================================================
N_SPLITS = 4
cv_folds = []
chunk_size = len(df_cv_pool) // (N_SPLITS + 1)

print(f"📦 Generando {N_SPLITS} folds de validación temporal...")

for fold in range(N_SPLITS):
    # Enfoque expansivo cronológico
    train_end_idx = chunk_size * (fold + 1)
    val_end_idx = train_end_idx + chunk_size

    train_fold = df_cv_pool.slice(0, train_end_idx)
    val_fold = df_cv_pool.slice(train_end_idx, chunk_size)

    X_tr_raw = train_fold.select(feature_cols).to_numpy()
    y_tr = train_fold.select(target_col).to_numpy().squeeze().astype(int)

    X_va_raw = val_fold.select(feature_cols).to_numpy()
    y_va = val_fold.select(target_col).to_numpy().squeeze().astype(int)

    # Escalado Z-Score aislado por Fold
    scaler = StandardScaler()
    X_tr = scaler.fit_transform(X_tr_raw)
    X_va = scaler.transform(X_va_raw)

    # Pesos por desbalanceo de este fold
    class_counts = np.bincount(y_tr, minlength=3)
    class_weights = len(y_tr) / (3.0 * np.where(class_counts == 0, 1, class_counts))
    weights_t = torch.tensor(class_weights, dtype=torch.float32).to(device)

    # Pasamos los tensores directamente a la GPU (Acelera x10 el bucle de Hyperopt)
    cv_folds.append({
        'X_train_t': torch.tensor(X_tr, dtype=torch.float32).to(device),
        'y_train_t': torch.tensor(y_tr, dtype=torch.long).to(device),
        'X_val_t': torch.tensor(X_va, dtype=torch.float32).to(device),
        'y_val_true': y_va,
        'weights_t': weights_t
    })

# Matrices del Test Ciego Final
scaler_final = StandardScaler()
X_cv_full_raw = df_cv_pool.select(feature_cols).to_numpy()
y_cv_full = df_cv_pool.select(target_col).to_numpy().squeeze().astype(int)

X_train_full = scaler_final.fit_transform(X_cv_full_raw)
X_test_final = scaler_final.transform(df_final_test.select(feature_cols).to_numpy())
y_test_final = df_final_test.select(target_col).to_numpy().squeeze().astype(int)

class_counts_f = np.bincount(y_cv_full, minlength=3)
class_weights_f = len(y_cv_full) / (3.0 * class_counts_f)
class_weights_final_t = torch.tensor(class_weights_f, dtype=torch.float32).to(device)

# =====================================================================
# 4. DEFINICIÓN DE LA ARQUITECTURA DINÁMICA
# =====================================================================
class DynamicMLP(nn.Module):
    def __init__(self, input_dim, layers_config, dropout_rate):
        super(DynamicMLP, self).__init__()
        layers = []
        prev_dim = input_dim

        for units in layers_config:
            layers.append(nn.Linear(prev_dim, units))
            layers.append(nn.BatchNorm1d(units))
            layers.append(nn.Hardswish())
            layers.append(nn.Dropout(dropout_rate))
            prev_dim = units

        layers.append(nn.Linear(prev_dim, 3))
        self.network = nn.Sequential(*layers)
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='relu')
                nn.init.constant_(m.bias, 0.0)

    def forward(self, x):
        return self.network(x)

# =====================================================================
# 5. ESPACIO DE BÚSQUEDA Y FUNCIÓN OBJETIVO CON VALIDACIÓN CRUZADA
# =====================================================================
space_mlp = {
    'architecture': hp.choice('architecture', [
        [256, 128, 64],
        [512, 256, 64],
        [128, 64],
        [512, 128]
    ]),
    'lr': hp.loguniform('lr', np.log(1e-4), np.log(1e-2)),
    'dropout': hp.uniform('dropout', 0.1, 0.5),
    'batch_size': hp.choice('batch_size', [512, 1024, 2048])
}

def objective_mlp(params):
    f1_scores_folds = []

    # Iteramos sobre cada ventana temporal
    for fold in cv_folds:
        train_dataset = TensorDataset(fold['X_train_t'], fold['y_train_t'])
        train_loader = DataLoader(train_dataset, batch_size=params['batch_size'], shuffle=True)

        model = DynamicMLP(
            input_dim=fold['X_train_t'].shape[1],
            layers_config=params['architecture'],
            dropout_rate=params['dropout']
        ).to(device)

        criterion = nn.CrossEntropyLoss(weight=fold['weights_t'])
        optimizer = optim.AdamW(model.parameters(), lr=params['lr'], weight_decay=1e-3)

        # Entrenamos 5 épocas por fold (rápido en GPU porque los datos ya residen allí)
        for epoch in range(5):
            model.train()
            for batch_x, batch_y in train_loader:
                optimizer.zero_grad()
                outputs = model(batch_x)
                loss = criterion(outputs, batch_y)
                if torch.isnan(loss) or torch.isinf(loss):
                    return {'loss': 999, 'status': STATUS_OK}
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

        # EVALUACIÓN REAL: En el set de Validación del fold (Datos que NO vio el modelo)
        model.eval()
        with torch.no_grad():
            logits = model(fold['X_val_t'])
            preds = torch.argmax(logits, dim=1).cpu().numpy()

        # Medimos F1 Macro en las colas (0 y 2) para evaluar la robustez de la predicción
        score = f1_score(fold['y_val_true'], preds, labels=[0, 2], average='macro', zero_division=0)
        f1_scores_folds.append(score)

    mean_f1_cv = np.mean(f1_scores_folds)
    return {'loss': -mean_f1_cv, 'status': STATUS_OK}

# =====================================================================
# 6. TUNING BAYESIANO TEMPORAL
# =====================================================================
print("\n🕵️ Buscando la arquitectura óptima mediante Validación Cruzada Temporal...")
trials_mlp = Trials()
best_mlp_idx = fmin(
    fn=objective_mlp,
    space=space_mlp,
    algo=tpe.suggest,
    max_evals=12,  # 12 evaluaciones cruzadas temporales robustas
    trials=trials_mlp
)

architectures_pool = [[256, 128, 64], [512, 256, 64], [128, 64], [512, 128]]
batch_sizes_pool = [512, 1024, 2048]

best_architecture = architectures_pool[best_mlp_idx['architecture']]
best_batch_size = batch_sizes_pool[best_mlp_idx['batch_size']]
best_lr = best_mlp_idx['lr']
best_dropout = best_mlp_idx['dropout']

print("\n🏆 Configuración óptima multi-régimen encontrada:")
print(f"   - Arquitectura de capas: {best_architecture}")
print(f"   - Learning Rate: {best_lr:.6f}")
print(f"   - Tasa de Dropout: {best_dropout:.2f}")
print(f"   - Tamaño de lote: {best_batch_size}")

# =====================================================================
# 7. ENTRENAMIENTO FINAL COMPLETO (Sobre todo el histórico agrupado)
# =====================================================================
print("\n🔥 Entrenando el MLP final optimizado...")
X_train_full_t = torch.tensor(X_train_full, dtype=torch.float32)
y_train_full_t = torch.tensor(y_cv_full, dtype=torch.long)

train_dataset = TensorDataset(X_train_full_t, y_train_full_t)
final_loader = DataLoader(train_dataset, batch_size=best_batch_size, shuffle=True)

final_model = DynamicMLP(X_train_full.shape[1], best_architecture, best_dropout).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights_final_t)
optimizer = optim.AdamW(final_model.parameters(), lr=best_lr, weight_decay=1e-3)

for epoch in range(15):
    final_model.train()
    running_loss = 0.0
    for batch_x, batch_y in final_loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        optimizer.zero_grad()
        outputs = final_model(batch_x)
        loss = criterion(outputs, batch_y)
        if torch.isnan(loss): continue
        loss.backward()
        nn.utils.clip_norm = nn.utils.clip_grad_norm_(final_model.parameters(), max_norm=1.0)
        optimizer.step()
        running_loss += loss.item() * batch_x.size(0)
    print(f"Época Final {epoch+1:02d}/15 | Loss: {running_loss/len(y_cv_full):.4f}")

# =====================================================================
# 8. INFERENCIA Y EVALUACIÓN MULTI-UMBRAL EN TEST CIEGO REAL
# =====================================================================
final_model.eval()
X_test_final_t = torch.tensor(X_test_final, dtype=torch.float32).to(device)
with torch.no_grad():
    logits = final_model(X_test_final_t)
    preds_prob_mlp = nn.functional.softmax(logits, dim=1).cpu().numpy()

umbrales_a_testear = [0.45, 0.50, 0.55, 0.60, 0.65, 0.70]
prob_down = preds_prob_mlp[:, 0]
prob_up = preds_prob_mlp[:, 2]

for ts in umbrales_a_testear:
    print(f"\n" + "="*70)
    print(f"📊 REPORT MULTI-REGIMEN MLP - UMBRAL DEL {int(ts*100)}% (TEST CIEGO OUT-OF-SAMPLE)")
    print("="*70)

    y_pred_ts = np.ones(len(preds_prob_mlp), dtype=int)
    y_pred_ts[prob_down >= ts] = 0
    y_pred_ts[prob_up >= ts] = 2

    print(classification_report(y_test_final, y_pred_ts, target_names=["DOWN (0)", "STAT (1)", "UP (2)"], zero_division=0))

    cm = confusion_matrix(y_test_final, y_pred_ts)
    print("Matriz de Confusión:")
    print(f"{'':<15} Pred: DOWN   Pred: STAT   Pred: UP")
    print(f"Real: DOWN     {cm[0][0]:<12}{cm[0][1]:<13}{cm[0][2]}")
    print(f"Real: STAT     {cm[1][0]:<12}{cm[1][1]:<13}{cm[1][2]}")
    print(f"Real: UP       {cm[2][0]:<12}{cm[2][1]:<13}{cm[2][2]}")

    total_trades = np.sum((y_pred_ts == 0) | (y_pred_ts == 2))
    pct_operado = (total_trades / len(y_test_final)) * 100
    print(f"📈 Grado de actividad en mercado: {pct_operado:.2f}% de los ticks operados.")
```
